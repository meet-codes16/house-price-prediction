```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

ct = ColumnTransformer(
    transformers=[('encoder', OneHotEncoder(), ['city'])],
    remainder='passthrough'  
)

x= ct.fit_transform(x) 
```

```python
x=x.toarray()
```

```python
print(x)
```

```python
y = y.values.reshape(len(y), 1)
```

```python
print(y)
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 1)
```

```python
from sklearn.preprocessing import StandardScaler
sc_x=StandardScaler()
sc_y=StandardScaler()
x_train=sc_x.fit_transform(x_train)
y_train=sc_y.fit_transform(y_train)
x_test=sc_x.transform(x_test)
y_test=sc_y.transform(y_test)
```

```python
from sklearn.svm import SVR
regressor=SVR(kernel='rbf')
regressor.fit(x_train,y_train)
```

```python
y_pred=sc_y.inverse_transform(regressor.predict(x_train).reshape(-1,1))
```

```python
from sklearn.metrics import r2_score
r2_score(sc_y.inverse_transform(y_train),y_pred)
```

```python
import matplotlib.pyplot as plt

residuals = sc_y.inverse_transform(y_train)- y_pred

plt.scatter(y_pred, residuals)  # X = predicted, Y = error
plt.axhline(0, color='red', linestyle='--')  # baseline = perfect prediction
plt.xlabel("Predicted values")
plt.ylabel("Residuals")
plt.show()
```
